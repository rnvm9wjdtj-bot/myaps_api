```python
db = ['haida1', 'haida2', 'haida3', 'haida4', 'dev', 'dev_mes']

tables = ['t_planner', 't_groupcolor', 't_setuptime', 't_shiftdate', 't_shifttime', 't_material', 't_workcenter', 't_mat_ver', 't_mat_wc', 't_mat_wc_bom', 't_mold', 't_mat_wc_mold', 't_demand', 't_supply', 't_orderwc', 't_confirm', 't_peg']

views = ['v_supply_mo', 'v_orderwc', 'v_matdailyqtyreport', 'v_confirm', 'v_supply_complete']

user = 'myaps'

password = 'Myaps@Haida'

grant_sql = f"DROP USER IF EXISTS '{user}'@'%';\nFLUSH PRIVILEGES;\nCREATE USER '{user}'@'%' IDENTIFIED BY '{password}';\nFLUSH PRIVILEGES;\n"

for db_name in db:
    for table_name in tables:
        grant_sql += f"GRANT SELECT, INSERT, UPDATE, DELETE ON {db_name}.{table_name} TO '{user}'@'%';\n"
    for view_name in views:
        grant_sql += f"GRANT SELECT ON {db_name}.{view_name} TO '{user}'@'%';\n"

    # 授予用户所有存储过程和函数的执行权限
    grant_sql += f"GRANT EXECUTE ON {db_name}.* TO '{user}'@'%';\n"

# 授予用户binlog权限
grant_sql += f"GRANT REPLICATION SLAVE, REPLICATION CLIENT ON *.* TO '{user}'@'%';\nFLUSH PRIVILEGES;"

print(grant_sql)
```
